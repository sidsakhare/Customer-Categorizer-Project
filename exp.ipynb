```python
import sys
from typing import Optional
import numpy as np
import pandas as pd
from pandas import DataFrame

from src.configuration.mongodb_connection import MongoDBClient
from src.constants.database import DATABASE_NAME, COLLECTION_NAME
from src.exception import  CustomException

from dotenv import load_dotenv
load_dotenv(r"C:\Users\ACER\OneDrive\Desktop\ML Notes and colab Notebooks\Data Science Projects\Customer-Categorization-new\.env")

class CustomerData:
    """
    This class help to export entire mongo db record as pandas dataframe
    """
    def __init__(self):
        try:
            self.mongo_client = MongoDBClient(database_name= DATABASE_NAME)
        except Exception as e:
            raise CustomException(e,sys) from e
    
    def export_collection_as_dataframe(
        self, collection_name:str, database_name:Optional[str]
    ) -> DataFrame:
        try:
            '''
            export entire collectin as dataframe:
            return pd.DataFrame of collection
            '''
            if database_name is None:
                collection = self.mongo_client.database[collection_name]
            else:
                collection = self.mongo_client.client[database_name][collection_name]
            df = pd.DataFrame(list(collection.find()))
            if "_id" in df.columns.to_list():
                df.drop(columns = ["_id"], inplace= True)
            df.replace({"na":np.nan}, inplace = True)
            return df
        except Exception as e:
            raise CustomException(e,sys) from e



```

```python
from src.data_access.customer_data import CustomerData  # adjust import path to wherever this class lives

# 1. Instantiate — this connects to Mongo via MongoDBClient internally
customer_data = CustomerData()

# 2a. Export a collection from the DEFAULT database (the one set in DATABASE_NAME)
df = customer_data.export_collection_as_dataframe(
    collection_name= COLLECTION_NAME
)

print(df.shape)
print(df.head())
```
